```python
def classify(x):
    string = ''
    for i in range(0, len(x)):
        if ((i == 0) and (65 <= ord(x[i]) <= 90)) or (((ord(x[i - 1]) == 32) or (ord(x[i - 1]) == 45)) and (65 <= ord(x[i]) <= 90)):
            string += x[i]
        elif ((i == 0) and (97 <= ord(x[i]) <= 122)) or (((ord(x[i - 1]) == 32) or (ord(x[i - 1]) == 45)) and (97 <= ord(x[i]) <= 122)):
            string += chr(ord(x[i]) - 32)
        else:
            string += x[i]
    return string
```

```python
df['area_type'] = df['area_type'].apply(classify)
```

```python
def classify4(x):
    if '-' in x:
        for i in range(0, len(x)):
            if x[i] == '-':
                return x[0 : i - 1]
            else:
                continue
    else:
        return x
```

```python
df['sqft1'] = df['total_sqft'].apply(classify4)
```

```python
def classify5(x):
    if '-' in x:
        for i in range(0, len(x)):
            if x[i] == '-':
                return x[i + 2 : len(x)]
            else:
                continue
    else:
        return x
```

```python
df['sqft2'] = df['total_sqft'].apply(classify5)
```

```python
#Sq. Yards (1 * 9)
#Sq. Meter (1 * 10.76)
#Acres (1 * 43560)
#Guntha (1 * 1089)
#Grounds (1 * 2400)
#Cents (1 * 435.6)
```

```python
def calculation(x):
    for i in range(0, len(x)):
        if 'Yards' in x:
            if x[i] == 'S':
                return float(x[0 : i]) * 9
        elif 'Meter' in x: 
            if x[i] == 'S':
                return float(x[0 : i]) * 10.76
        elif 'Acres' in x:
            if x[i] == 'A':
                return float(x[0 : i]) * 43560
        elif 'Grounds' in x:
            if x[i] == 'G':
                return float(x[0 : i]) * 2400
        elif 'Guntha' in x:
            if x[i] == 'G':
                return float(x[0 : i]) * 1089
        elif 'Cents' in x:
            if x[i] == 'C':
                return float(x[0 : i]) * 435.6
        elif 'Perch' in x:
            if x[i] == 'P':
                return float(x[0 : i]) * 272.25
        else:
            return float(x)
```

```python
df['sqft1'] = df['sqft1'].apply(calculation)
```
